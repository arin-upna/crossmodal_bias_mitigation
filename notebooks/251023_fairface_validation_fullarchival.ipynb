## General performance analysis

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.patches import Rectangle
from matplotlib.collections import PatchCollection

# Set up the plotting style
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (16, 12)

# Define color scheme
cmap = sns.color_palette("magma_r", as_cmap=True)
border_color = 'darkblue'
border_width = 3

# Define the correct mappings for each prediction type
correct_mappings = {
    'gender': {
        'Female': ['female'],
        'Male': ['male']
    },
    'race_to_race3': {
        'White': ['Caucasian'],
        'Southeast Asian': ['Asian'],
        'East Asian': ['Asian'],
        'Black': ['African-American'],
        'Latino_Hispanic': [],  # No direct mapping
        'Middle Eastern': [],   # No direct mapping
        'Indian': []            # No direct mapping
    },
    'race4_to_race3': {
        'White': ['Caucasian'],
        'Asian': ['Asian'],
        'Black': ['African-American'],
        'Indian': []  # Always incorrect as you mentioned
    },
    'age': {
        '0-2': ['0-3'],
        '3-9': ['4-19'],
        '10-19': ['4-19'],
        '20-29': ['20-39'],
        '30-39': ['20-39'],
        '40-49': ['40-69'],
        '50-59': ['40-69'],
        '60-69': ['40-69'],
        '70+': ['70+']
    }
}

# Function to create normalized crosstab with proper ordering
def create_normalized_crosstab(df, pred_col, true_col, pred_order=None, true_order=None):
    # Remove nan values
    df_clean = df[[pred_col, true_col]].dropna()
    
    # Create crosstab
    ct = pd.crosstab(df_clean[true_col], df_clean[pred_col], margins=False)
    
    # Normalize by row (original demographic) to get percentages
    ct_normalized = ct.div(ct.sum(axis=1), axis=0) * 100
    
    # Reorder if orders are provided
    if pred_order:
        pred_order = [p for p in pred_order if p in ct_normalized.columns]
        ct_normalized = ct_normalized[pred_order]
    if true_order:
        true_order = [t for t in true_order if t in ct_normalized.index]
        ct_normalized = ct_normalized.reindex(true_order, fill_value=0)
    
    return ct_normalized, ct  # Return both normalized and raw counts

# Function to add borders to correct predictions
def add_borders(ax, crosstab, mapping, pred_col, true_col):
    patches = []
    
    for i, true_val in enumerate(crosstab.index):
        for j, pred_val in enumerate(crosstab.columns):
            if pred_val in mapping and true_val in mapping[pred_val]:
                # Add a rectangle border for correct predictions
                rect = Rectangle((j, i), 1, 1, 
                                linewidth=border_width, 
                                edgecolor=border_color,
                                facecolor='none',
                                zorder=10)
                ax.add_patch(rect)

# Define orderings
gender_order_pred = ['Female', 'Male']
gender_order_true = ['female', 'male', 'unsure']
ct_gender_norm, ct_gender_raw = create_normalized_crosstab(data_rafdb, 'gender', 'original_gender', 
                                                           gender_order_pred, gender_order_true)

race_order_pred = ['White', 'Black', 'East Asian', 'Southeast Asian', 
                   'Indian', 'Latino_Hispanic', 'Middle Eastern']
race_order_true = ['Caucasian', 'African-American', 'Asian']
ct_race_norm, ct_race_raw = create_normalized_crosstab(data_rafdb, 'race', 'original_race',
                                                        race_order_pred, race_order_true)

race4_order_pred = ['White', 'Black', 'Asian', 'Indian']
ct_race4_norm, ct_race4_raw = create_normalized_crosstab(data_rafdb, 'race4', 'original_race',
                                                          race4_order_pred, race_order_true)

age_order_pred = ['0-2', '3-9', '10-19', '20-29', '30-39', '40-49', '50-59', '60-69', '70+']
age_order_true = ['0-3', '4-19', '20-39', '40-69', '70+']
ct_age_norm, ct_age_raw = create_normalized_crosstab(data_rafdb, 'age', 'original_age',
                                                      age_order_pred, age_order_true)

# Create the figure
fig, axes = plt.subplots(2, 2, figsize=(16, 12))
fig.suptitle('FairFace Prediction Confusion Matrices (Normalized by Original Demographics)', 
             fontsize=16, fontweight='bold')

# Custom annotation function to show both percentage and count
def custom_annot(data_norm, data_raw):
    annot = []
    for i in range(len(data_norm.index)):
        row = []
        for j in range(len(data_norm.columns)):
            pct = data_norm.iloc[i, j]
            cnt = data_raw.iloc[i, j] if i < len(data_raw.index) and j < len(data_raw.columns) else 0
            if pct > 0:
                row.append(f'{pct:.1f}%\n({cnt:.0f})')
            else:
                row.append('')
        annot.append(row)
    return annot

# 1. Gender vs Original Gender
ax1 = axes[0, 0]
annot_gender = custom_annot(ct_gender_norm, ct_gender_raw)
sns.heatmap(ct_gender_norm, annot=annot_gender, fmt='', cmap=cmap, ax=ax1, 
            cbar_kws={'label': 'Percentage (%)'}, vmin=0, vmax=100)
add_borders(ax1, ct_gender_norm, correct_mappings['gender'], 'gender', 'original_gender')
ax1.set_title('Gender Predictions vs Original Gender (% by row)', fontsize=14, fontweight='bold')
ax1.set_xlabel('Predicted Gender', fontsize=12)
ax1.set_ylabel('Original Gender', fontsize=12)

# Add row totals
for i, val in enumerate(ct_gender_raw.sum(axis=1)):
    ax1.text(-0.5, i + 0.5, f'n={val:.0f}', ha='right', va='center', fontweight='bold', fontsize=10)

# 2. Race vs Original Race
ax2 = axes[1, 1]
annot_race = custom_annot(ct_race_norm, ct_race_raw)
sns.heatmap(ct_race_norm, annot=annot_race, fmt='', cmap=cmap, ax=ax2,
            cbar_kws={'label': 'Percentage (%)'}, vmin=0, vmax=100)
add_borders(ax2, ct_race_norm, correct_mappings['race_to_race3'], 'race', 'original_race')
ax2.set_title('Race Predictions (7 classes) vs Original Race (% by row)', fontsize=14, fontweight='bold')
ax2.set_xlabel('Predicted Race', fontsize=12)
ax2.set_ylabel('Original Race', fontsize=12)
ax2.tick_params(axis='x', rotation=45)

# Add row totals
for i, val in enumerate(ct_race_raw.sum(axis=1)):
    ax2.text(-0.5, i + 0.5, f'n={val:.0f}', ha='right', va='center', fontweight='bold', fontsize=10)

# 3. Race4 vs Original Race
ax3 = axes[1, 0]
annot_race4 = custom_annot(ct_race4_norm, ct_race4_raw)
sns.heatmap(ct_race4_norm, annot=annot_race4, fmt='', cmap=cmap, ax=ax3,
            cbar_kws={'label': 'Percentage (%)'}, vmin=0, vmax=100)
add_borders(ax3, ct_race4_norm, correct_mappings['race4_to_race3'], 'race4', 'original_race')
ax3.set_title('Race4 Predictions (4 classes) vs Original Race (% by row)', fontsize=14, fontweight='bold')
ax3.set_xlabel('Predicted Race (4 classes)', fontsize=12)
ax3.set_ylabel('Original Race', fontsize=12)

# Add row totals
for i, val in enumerate(ct_race4_raw.sum(axis=1)):
    ax3.text(-0.5, i + 0.5, f'n={val:.0f}', ha='right', va='center', fontweight='bold', fontsize=10)

# 4. Age vs Original Age
ax4 = axes[0, 1]
annot_age = custom_annot(ct_age_norm, ct_age_raw)
sns.heatmap(ct_age_norm, annot=annot_age, fmt='', cmap=cmap, ax=ax4,
            cbar_kws={'label': 'Percentage (%)'}, vmin=0, vmax=100)
add_borders(ax4, ct_age_norm, correct_mappings['age'], 'age', 'original_age')
ax4.set_title('Age Predictions vs Original Age (% by row)', fontsize=14, fontweight='bold')
ax4.set_xlabel('Predicted Age Range', fontsize=12)
ax4.set_ylabel('Original Age Range', fontsize=12)

# Add row totals
for i, val in enumerate(ct_age_raw.sum(axis=1)):
    ax4.text(-0.5, i + 0.5, f'n={val:.0f}', ha='right', va='center', fontweight='bold', fontsize=10)

# Adjust layout
plt.tight_layout()
plt.show()
```

## Cross analysis (per emotion)

```python
# Calculate recall for each demographic group and emotion label combination
def calculate_recall_by_group_and_emotion(df, demo_col, original_demo_col, mapping):
    """
    Calculate recall (correct predictions / total) for each demographic group and emotion label.
    """
    # Get unique emotion labels
    emotions = df['label'].unique()
    emotions = sorted(emotions)
    
    # Initialize results dictionary
    recall_matrix = {}
    count_matrix = {}
    
    # For each original demographic group
    for orig_group in df[original_demo_col].dropna().unique():
        recall_matrix[orig_group] = {}
        count_matrix[orig_group] = {}
        
        # For each emotion
        for emotion in emotions:
            # Filter data for this group and emotion
            mask = (df[original_demo_col] == orig_group) & (df['label'] == emotion)
            subset = df[mask]
            
            total = len(subset)
            count_matrix[orig_group][emotion] = total
            
            if total > 0:
                # Count correct predictions based on mapping
                correct = 0
                for pred_group, orig_groups in mapping.items():
                    if orig_group in orig_groups:
                        correct += len(subset[subset[demo_col] == pred_group])
                
                recall_matrix[orig_group][emotion] = (correct / total) * 100
            else:
                recall_matrix[orig_group][emotion] = np.nan
    
    # Convert to DataFrames
    recall_df = pd.DataFrame(recall_matrix).T
    count_df = pd.DataFrame(count_matrix).T
    
    return recall_df, count_df

# Calculate recall for each demographic dimension
emotions_order = ['angry', 'disgust', 'fear', 'happy', 'neutral', 'sad', 'surprise']

# Gender recall
gender_recall, gender_counts = calculate_recall_by_group_and_emotion(
    data_rafdb, 'gender', 'original_gender', correct_mappings['gender']
)
gender_recall = gender_recall.reindex(gender_order_true, axis=0).reindex(emotions_order, axis=1)
gender_counts = gender_counts.reindex(gender_order_true, axis=0).reindex(emotions_order, axis=1)

# Race recall (7 classes to 3 classes)
race_recall, race_counts = calculate_recall_by_group_and_emotion(
    data_rafdb, 'race', 'original_race', correct_mappings['race_to_race3']
)
race_recall = race_recall.reindex(race_order_true, axis=0).reindex(emotions_order, axis=1)
race_counts = race_counts.reindex(race_order_true, axis=0).reindex(emotions_order, axis=1)

# Race4 recall
race4_recall, race4_counts = calculate_recall_by_group_and_emotion(
    data_rafdb, 'race4', 'original_race', correct_mappings['race4_to_race3']
)
race4_recall = race4_recall.reindex(race_order_true, axis=0).reindex(emotions_order, axis=1)
race4_counts = race4_counts.reindex(race_order_true, axis=0).reindex(emotions_order, axis=1)

# Age recall
age_recall, age_counts = calculate_recall_by_group_and_emotion(
    data_rafdb, 'age', 'original_age', correct_mappings['age']
)
age_recall = age_recall.reindex(age_order_true, axis=0).reindex(emotions_order, axis=1)
age_counts = age_counts.reindex(age_order_true, axis=0).reindex(emotions_order, axis=1)

# Custom annotation function for recall heatmaps
def create_recall_annot(recall_df, count_df):
    annot = []
    for i in range(len(recall_df.index)):
        row = []
        for j in range(len(recall_df.columns)):
            recall = recall_df.iloc[i, j]
            count = count_df.iloc[i, j]
            if pd.notna(recall) and count > 0:
                row.append(f'{recall:.1f}%\n(n={count:.0f})')
            else:
                row.append('N/A\n(n=0)')
        annot.append(row)
    return annot

# Create separate figures for each demographic dimension

# 1. Gender Recall Heatmap
fig1, ax1 = plt.subplots(figsize=(12, 3 + len(gender_recall) * 0.4))
annot_gender_recall = create_recall_annot(gender_recall, gender_counts)
sns.heatmap(gender_recall, annot=annot_gender_recall, fmt='', cmap=cmap, 
            cbar_kws={'label': 'Recall (%)'}, vmin=0, vmax=100, ax=ax1,
            linewidths=0.5, linecolor='gray')
ax1.set_title('Gender Classification Recall by Emotion Label', fontsize=14, fontweight='bold')
ax1.set_xlabel('Emotion Label', fontsize=12)
ax1.set_ylabel('Original Gender', fontsize=12)
ax1.set_xticklabels(ax1.get_xticklabels(), rotation=45, ha='right')

# Add row averages
for i, group in enumerate(gender_recall.index):
    mean_recall = gender_recall.loc[group].mean()
    if pd.notna(mean_recall):
        ax1.text(len(emotions_order) + 0.1, i + 0.5, f'Avg: {mean_recall:.1f}%', 
                va='center', fontweight='bold', fontsize=10)

plt.tight_layout()
plt.show()

# 2. Race (7-class) Recall Heatmap
fig2, ax2 = plt.subplots(figsize=(12, 3 + len(race_recall) * 0.4))
annot_race_recall = create_recall_annot(race_recall, race_counts)
sns.heatmap(race_recall, annot=annot_race_recall, fmt='', cmap=cmap,
            cbar_kws={'label': 'Recall (%)'}, vmin=0, vmax=100, ax=ax2,
            linewidths=0.5, linecolor='gray')
ax2.set_title('Race Classification Recall by Emotion Label (7 predicted classes → 3 original)', 
              fontsize=14, fontweight='bold')
ax2.set_xlabel('Emotion Label', fontsize=12)
ax2.set_ylabel('Original Race', fontsize=12)
ax2.set_xticklabels(ax2.get_xticklabels(), rotation=45, ha='right')

# Add row averages
for i, group in enumerate(race_recall.index):
    mean_recall = race_recall.loc[group].mean()
    if pd.notna(mean_recall):
        ax2.text(len(emotions_order) + 0.1, i + 0.5, f'Avg: {mean_recall:.1f}%', 
                va='center', fontweight='bold', fontsize=10)

plt.tight_layout()
plt.show()

# 3. Race4 Recall Heatmap
fig3, ax3 = plt.subplots(figsize=(12, 3 + len(race4_recall) * 0.4))
annot_race4_recall = create_recall_annot(race4_recall, race4_counts)
sns.heatmap(race4_recall, annot=annot_race4_recall, fmt='', cmap=cmap,
            cbar_kws={'label': 'Recall (%)'}, vmin=0, vmax=100, ax=ax3,
            linewidths=0.5, linecolor='gray')
ax3.set_title('Race4 Classification Recall by Emotion Label (4 predicted classes → 3 original)', 
              fontsize=14, fontweight='bold')
ax3.set_xlabel('Emotion Label', fontsize=12)
ax3.set_ylabel('Original Race', fontsize=12)
ax3.set_xticklabels(ax3.get_xticklabels(), rotation=45, ha='right')

# Add row averages
for i, group in enumerate(race4_recall.index):
    mean_recall = race4_recall.loc[group].mean()
    if pd.notna(mean_recall):
        ax3.text(len(emotions_order) + 0.1, i + 0.5, f'Avg: {mean_recall:.1f}%', 
                va='center', fontweight='bold', fontsize=10)

plt.tight_layout()
plt.show()

# 4. Age Recall Heatmap
fig4, ax4 = plt.subplots(figsize=(12, 3 + len(age_recall) * 0.4))
annot_age_recall = create_recall_annot(age_recall, age_counts)
sns.heatmap(age_recall, annot=annot_age_recall, fmt='', cmap=cmap,
            cbar_kws={'label': 'Recall (%)'}, vmin=0, vmax=100, ax=ax4,
            linewidths=0.5, linecolor='gray')
ax4.set_title('Age Classification Recall by Emotion Label', fontsize=14, fontweight='bold')
ax4.set_xlabel('Emotion Label', fontsize=12)
ax4.set_ylabel('Original Age', fontsize=12)
ax4.set_xticklabels(ax4.get_xticklabels(), rotation=45, ha='right')

# Add row averages
for i, group in enumerate(age_recall.index):
    mean_recall = age_recall.loc[group].mean()
    if pd.notna(mean_recall):
        ax4.text(len(emotions_order) + 0.1, i + 0.5, f'Avg: {mean_recall:.1f}%', 
                va='center', fontweight='bold', fontsize=10)

plt.tight_layout()
plt.show()

# Print summary statistics
print("\n" + "="*60)
print("DEMOGRAPHIC CLASSIFICATION RECALL SUMMARY BY EMOTION")
print("="*60)

# Function to find best and worst performing emotion for each group
def print_summary_stats(recall_df, demo_name):
    print(f"\n--- {demo_name.upper()} ---")
    for group in recall_df.index:
        if group != 'unsure':  # Skip unsure if present
            recalls = recall_df.loc[group]
            valid_recalls = recalls.dropna()
            if len(valid_recalls) > 0:
                best_emotion = valid_recalls.idxmax()
                worst_emotion = valid_recalls.idxmin()
                print(f"\n{group}:")
                print(f"  Best:  {best_emotion:8s} = {valid_recalls[best_emotion]:.1f}%")
                print(f"  Worst: {worst_emotion:8s} = {valid_recalls[worst_emotion]:.1f}%")
                print(f"  Range: {valid_recalls.max() - valid_recalls.min():.1f}% difference")
                print(f"  Avg:   {valid_recalls.mean():.1f}%")

print_summary_stats(gender_recall, "Gender")
print_summary_stats(race_recall, "Race (7-class)")
print_summary_stats(race4_recall, "Race4 (4-class)")
print_summary_stats(age_recall, "Age")

# Calculate overall variance across emotions for each demographic
print("\n" + "="*60)
print("VARIANCE IN RECALL ACROSS EMOTIONS")
print("="*60)

for name, df in [("Gender", gender_recall), ("Race", race_recall), 
                  ("Race4", race4_recall), ("Age", age_recall)]:
    variances = df.var(axis=1)
    print(f"\n{name} - Standard deviation of recall across emotions:")
    for group in variances.index:
        if pd.notna(variances[group]):
            print(f"  {group:20s}: {np.sqrt(variances[group]):.1f}%")
```

```python
# First, let's explore the FACES dataset structure
print("FACES Dataset Overview:")
print(f"Total samples: {len(data_faces)}")
print(f"Unique original ages: {sorted(data_faces['original_age'].dropna().unique())}")
print(f"Age range: {data_faces['original_age'].min()} - {data_faces['original_age'].max()}")

# Create age groups from original_age for better analysis
def categorize_original_age(age):
    if pd.isna(age):
        return None
    age = int(age)
    if age < 20:
        return '10-19'
    elif age < 30:
        return '20-29'
    elif age < 40:
        return '30-39'
    elif age < 50:
        return '40-49'
    elif age < 60:
        return '50-59'
    elif age < 70:
        return '60-69'
    else:
        return '70+'

data_faces['original_age_group'] = data_faces['original_age'].apply(categorize_original_age)

# Create normalized gender confusion matrix for FACES
gender_order_pred_faces = ['Female', 'Male']
gender_order_true_faces = ['female', 'male']

ct_gender_norm_faces, ct_gender_raw_faces = create_normalized_crosstab(
    data_faces, 'gender', 'original_gender', 
    gender_order_pred_faces, gender_order_true_faces
)

# Create the figure for FACES gender analysis
fig, ax = plt.subplots(figsize=(8, 6))

# Custom annotation function (reuse from before)
annot_gender_faces = custom_annot(ct_gender_norm_faces, ct_gender_raw_faces)

# Create heatmap
sns.heatmap(ct_gender_norm_faces, annot=annot_gender_faces, fmt='', cmap=cmap, ax=ax, 
            cbar_kws={'label': 'Percentage (%)'}, vmin=0, vmax=100,
            linewidths=0.5, linecolor='gray')

# Add borders for correct predictions
add_borders(ax, ct_gender_norm_faces, correct_mappings['gender'], 'gender', 'original_gender')

ax.set_title('FACES Dataset: Gender Predictions vs Original Gender (% by row)', 
             fontsize=14, fontweight='bold')
ax.set_xlabel('Predicted Gender (FairFace)', fontsize=12)
ax.set_ylabel('Original Gender', fontsize=12)

# Add row totals
for i, val in enumerate(ct_gender_raw_faces.sum(axis=1)):
    ax.text(-0.5, i + 0.5, f'n={val:.0f}', ha='right', va='center', 
            fontweight='bold', fontsize=10)

plt.tight_layout()
plt.show()

# Print accuracy metrics
print("\n" + "="*50)
print("FACES DATASET - GENDER PREDICTION ACCURACY")
print("="*50)

for orig in gender_order_true_faces:
    if orig in ct_gender_norm_faces.index:
        if orig == 'female':
            acc = ct_gender_norm_faces.loc[orig, 'Female'] if 'Female' in ct_gender_norm_faces.columns else 0
            misclass = ct_gender_norm_faces.loc[orig, 'Male'] if 'Male' in ct_gender_norm_faces.columns else 0
            n = ct_gender_raw_faces.sum(axis=1).loc[orig]
            print(f"\nFemale (n={n:.0f}):")
            print(f"  Correctly classified as Female: {acc:.1f}%")
            print(f"  Misclassified as Male: {misclass:.1f}%")
        elif orig == 'male':
            acc = ct_gender_norm_faces.loc[orig, 'Male'] if 'Male' in ct_gender_norm_faces.columns else 0
            misclass = ct_gender_norm_faces.loc[orig, 'Female'] if 'Female' in ct_gender_norm_faces.columns else 0
            n = ct_gender_raw_faces.sum(axis=1).loc[orig]
            print(f"\nMale (n={n:.0f}):")
            print(f"  Correctly classified as Male: {acc:.1f}%")
            print(f"  Misclassified as Female: {misclass:.1f}%")

# Compare with RAF-DB results if available
print("\n" + "="*50)
print("COMPARISON: FACES vs RAF-DB")
print("="*50)

if 'ct_gender_norm' in locals():
    print("\nGender Classification Accuracy:")
    print("RAF-DB:")
    if 'female' in ct_gender_norm.index and 'Female' in ct_gender_norm.columns:
        print(f"  Female → Female: {ct_gender_norm.loc['female', 'Female']:.1f}%")
    if 'male' in ct_gender_norm.index and 'Male' in ct_gender_norm.columns:
        print(f"  Male → Male: {ct_gender_norm.loc['male', 'Male']:.1f}%")
    
    print("\nFACES:")
    if 'female' in ct_gender_norm_faces.index and 'Female' in ct_gender_norm_faces.columns:
        print(f"  Female → Female: {ct_gender_norm_faces.loc['female', 'Female']:.1f}%")
    if 'male' in ct_gender_norm_faces.index and 'Male' in ct_gender_norm_faces.columns:
        print(f"  Male → Male: {ct_gender_norm_faces.loc['male', 'Male']:.1f}%")

# Analyze by age groups to see if there's an age effect
print("\n" + "="*50)
print("FACES: GENDER ACCURACY BY AGE GROUP")
print("="*50)

for age_group in sorted(data_faces['original_age_group'].dropna().unique()):
    subset = data_faces[data_faces['original_age_group'] == age_group]
    
    # Female accuracy in this age group
    female_subset = subset[subset['original_gender'] == 'female']
    if len(female_subset) > 0:
        female_correct = len(female_subset[female_subset['gender'] == 'Female'])
        female_acc = (female_correct / len(female_subset)) * 100
    else:
        female_acc = None
    
    # Male accuracy in this age group
    male_subset = subset[subset['original_gender'] == 'male']
    if len(male_subset) > 0:
        male_correct = len(male_subset[male_subset['gender'] == 'Male'])
        male_acc = (male_correct / len(male_subset)) * 100
    else:
        male_acc = None
    
    print(f"\nAge {age_group}:")
    if female_acc is not None:
        print(f"  Female accuracy: {female_acc:.1f}% (n={len(female_subset)})")
    if male_acc is not None:
        print(f"  Male accuracy: {male_acc:.1f}% (n={len(male_subset)})")

# Check if there's any emotion-specific pattern
print("\n" + "="*50)
print("FACES: GENDER ACCURACY BY EMOTION")
print("="*50)

for emotion in sorted(data_faces['label'].dropna().unique()):
    subset = data_faces[data_faces['label'] == emotion]
    
    # Female accuracy for this emotion
    female_subset = subset[subset['original_gender'] == 'female']
    if len(female_subset) > 0:
        female_correct = len(female_subset[female_subset['gender'] == 'Female'])
        female_acc = (female_correct / len(female_subset)) * 100
    else:
        female_acc = None
    
    # Male accuracy for this emotion
    male_subset = subset[subset['original_gender'] == 'male']
    if len(male_subset) > 0:
        male_correct = len(male_subset[male_subset['gender'] == 'Male'])
        male_acc = (male_correct / len(male_subset)) * 100
    else:
        male_acc = None
    
    print(f"\n{emotion.capitalize()}:")
    if female_acc is not None:
        print(f"  Female accuracy: {female_acc:.1f}% (n={len(female_subset)})")
    if male_acc is not None:
        print(f"  Male accuracy: {male_acc:.1f}% (n={len(male_subset)})")
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output, HTML
from PIL import Image
import matplotlib.pyplot as plt
import matplotlib.patches as patches

def create_demographic_explorer(df, image_dir='/media/DATA/FER_data/raf-db/basic/Image/original'):
    """
    Create an interactive explorer for demographic classifications
    """
    
    # Get unique values for each demographic
    demo_options = {
        'original_gender': sorted(df['original_gender'].dropna().unique()),
        'original_age': sorted(df['original_age'].dropna().unique()) if 'original_age' in df.columns else [],
        'original_race': sorted(df['original_race'].dropna().unique()) if 'original_race' in df.columns else [],
        'gender': sorted(df['gender'].dropna().unique()),
        'age': sorted(df['age'].dropna().unique()),
        'race': sorted(df['race'].dropna().unique()),
        'race4': sorted(df['race4'].dropna().unique()),
        'label': sorted(df['label'].dropna().unique())
    }
    
    # Create dropdown widgets
    original_demo = widgets.Dropdown(
        options=['original_gender', 'original_race', 'original_age', 'label'],
        value='original_gender',
        description='Original:',
        style={'description_width': 'initial'}
    )
    
    original_value = widgets.Dropdown(
        options=demo_options['original_gender'],
        description='Value:',
        style={'description_width': 'initial'}
    )
    
    predicted_demo = widgets.Dropdown(
        options=['gender', 'race', 'race4', 'age'],
        value='gender',
        description='Predicted:',
        style={'description_width': 'initial'}
    )
    
    predicted_value = widgets.Dropdown(
        options=demo_options['gender'],
        description='Value:',
        style={'description_width': 'initial'}
    )
    
    n_samples = widgets.IntSlider(
        value=10,
        min=1,
        max=20,
        step=1,
        description='Samples:',
        style={'description_width': 'initial'}
    )
    
    dataset_select = widgets.Dropdown(
        options=['RAF-DB', 'FACES'],
        value='RAF-DB',
        description='Dataset:',
        style={'description_width': 'initial'}
    )
    
    show_button = widgets.Button(
        description='Show Examples',
        button_style='primary',
        icon='search'
    )
    
    output = widgets.Output()
    
    def update_original_values(*args):
        """Update dropdown values when original demographic changes"""
        original_value.options = demo_options.get(original_demo.value, [])
        if original_value.options:
            original_value.value = original_value.options[0]
    
    def update_predicted_values(*args):
        """Update dropdown values when predicted demographic changes"""
        predicted_value.options = demo_options.get(predicted_demo.value, [])
        if predicted_value.options:
            predicted_value.value = predicted_value.options[0]
    
    def on_dataset_change(*args):
        """Switch dataset when selection changes"""
        nonlocal df, demo_options
        
        if dataset_select.value == 'RAF-DB':
            df = data_rafdb
        else:
            df = data_faces
            
        # Update demo options for new dataset
        demo_options = {
            'original_gender': sorted(df['original_gender'].dropna().unique()),
            'original_age': sorted(df['original_age'].dropna().unique()) if 'original_age' in df.columns else [],
            'original_race': sorted(df['original_race'].dropna().unique()) if 'original_race' in df.columns else [],
            'gender': sorted(df['gender'].dropna().unique()),
            'age': sorted(df['age'].dropna().unique()),
            'race': sorted(df['race'].dropna().unique()),
            'race4': sorted(df['race4'].dropna().unique()),
            'label': sorted(df['label'].dropna().unique())
        }
        
        # Update dropdowns
        update_original_values()
        update_predicted_values()
    
    def show_examples(*args):
        """Display examples based on selection"""
        with output:
            clear_output(wait=True)
            
            # Get current dataset
            current_df = data_rafdb if dataset_select.value == 'RAF-DB' else data_faces
            
            # Filter data
            orig_col = original_demo.value
            pred_col = predicted_demo.value
            orig_val = original_value.value
            pred_val = predicted_value.value
            
            # Get subset matching original value
            subset = current_df[current_df[orig_col] == orig_val].copy()
            
            if len(subset) == 0:
                print(f"No samples found with {orig_col} = {orig_val}")
                return
            
            # Determine correct predictions based on mappings
            if orig_col.startswith('original_') and pred_col in ['gender', 'race', 'race4', 'age']:
                # Check if this is a demographic prediction case
                mapping_key = pred_col
                if pred_col == 'race' and orig_col == 'original_race':
                    mapping_key = 'race_to_race3'
                elif pred_col == 'race4' and orig_col == 'original_race':
                    mapping_key = 'race4_to_race3'
                
                if mapping_key in correct_mappings:
                    # Check if predicted value maps correctly to original value
                    correct_mask = subset.apply(
                        lambda row: orig_val in correct_mappings[mapping_key].get(row[pred_col], []),
                        axis=1
                    )
                else:
                    # Direct comparison
                    correct_mask = subset[pred_col] == pred_val
            else:
                # For non-demographic comparisons (like emotion labels)
                correct_mask = subset[pred_col] == pred_val
            
            correct_samples = subset[correct_mask]
            incorrect_samples = subset[~correct_mask]
            
            # Sample images
            n = n_samples.value
            correct_to_show = correct_samples.sample(min(n, len(correct_samples))) if len(correct_samples) > 0 else pd.DataFrame()
            incorrect_to_show = incorrect_samples.sample(min(n, len(incorrect_samples))) if len(incorrect_samples) > 0 else pd.DataFrame()
            
            # Display summary
            print("="*70)
            print(f"Dataset: {dataset_select.value}")
            print(f"Filter: {orig_col} = '{orig_val}'")
            print(f"Looking for: {pred_col} = '{pred_val}'")
            print("="*70)
            print(f"\nTotal samples with {orig_col} = '{orig_val}': {len(subset)}")
            print(f"Correctly classified as {pred_col} = '{pred_val}': {len(correct_samples)} ({100*len(correct_samples)/len(subset):.1f}%)")
            print(f"Incorrectly classified: {len(incorrect_samples)} ({100*len(incorrect_samples)/len(subset):.1f}%)")
            
            # Function to plot images
            def plot_image_grid(samples_df, title, border_color):
                if len(samples_df) == 0:
                    print(f"\nNo {title.lower()} to show")
                    return
                
                n_images = len(samples_df)
                cols = 5
                rows = (n_images + cols - 1) // cols
                
                fig, axes = plt.subplots(rows, cols, figsize=(15, 3*rows))
                if rows == 1:
                    axes = axes.reshape(1, -1)
                
                fig.suptitle(title, fontsize=14, fontweight='bold')
                
                for idx, (_, row) in enumerate(samples_df.iterrows()):
                    r, c = idx // cols, idx % cols
                    ax = axes[r, c]
                    
                    # Construct image path
                    if dataset_select.value == 'RAF-DB':
                        img_path = f"/media/DATA/FER_data/raf-db/basic/Image/original/{row['img_path']}"
                    else:
                        # For FACES, adjust path as needed
                        img_path = f"/media/DATA/FER_data/faces/FACES: A database of facial expressions in younger, middle-aged, and older women and men/{row['img_path']}"
                    
                    try:
                        img = Image.open(img_path)
                        ax.imshow(img)
                        
                        # Add border
                        rect = patches.Rectangle((0, 0), img.width-1, img.height-1, 
                                                linewidth=3, edgecolor=border_color, facecolor='none')
                        ax.add_patch(rect)
                        
                        # Add title with predictions
                        pred_text = f"{pred_col}: {row[pred_col]}"
                        other_info = []
                        
                        # Add other demographic predictions
                        for col in ['label', 'gender', 'race', 'race4', 'age']:
                            if col != pred_col and col in row and pd.notna(row[col]):
                                other_info.append(f"{col}: {row[col]}")
                        
                        ax.set_title(f"{pred_text}\n" + "\n".join(other_info[:3]), fontsize=8)
                        
                    except FileNotFoundError:
                        print("File not found:", img_path)
                        ax.text(0.5, 0.5, 'Image not found', ha='center', va='center')
                        ax.set_xlim(0, 1)
                        ax.set_ylim(0, 1)
                    
                    ax.axis('off')
                
                # Hide empty subplots
                for idx in range(n_images, rows * cols):
                    r, c = idx // cols, idx % cols
                    axes[r, c].axis('off')
                
                plt.tight_layout()
                plt.show()
            
            # Show correctly classified examples
            if len(correct_samples) > 0:
                print(f"\n{'-'*50}")
                print(f"CORRECTLY CLASSIFIED (showing {len(correct_to_show)} of {len(correct_samples)})")
                print(f"{'-'*50}")
                plot_image_grid(correct_to_show, "Correctly Classified", 'green')
            
            # Show incorrectly classified examples
            if len(incorrect_samples) > 0:
                print(f"\n{'-'*50}")
                print(f"INCORRECTLY CLASSIFIED (showing {len(incorrect_to_show)} of {len(incorrect_samples)})")
                print(f"{'-'*50}")
                
                # Show distribution of incorrect predictions
                incorrect_dist = incorrect_samples[pred_col].value_counts()
                print("\nIncorrect predictions distributed as:")
                for val, count in incorrect_dist.items():
                    print(f"  {val}: {count} ({100*count/len(incorrect_samples):.1f}%)")
                
                plot_image_grid(incorrect_to_show, "Incorrectly Classified", 'red')
    
    # Set up observers
    original_demo.observe(update_original_values, 'value')
    predicted_demo.observe(update_predicted_values, 'value')
    dataset_select.observe(on_dataset_change, 'value')
    show_button.on_click(show_examples)
    
    # Initial update
    update_original_values()
    update_predicted_values()
    
    # Layout
    controls = widgets.VBox([
        widgets.HTML("<h3>Demographic Classification Explorer</h3>"),
        widgets.HBox([dataset_select]),
        widgets.HTML("<b>Select demographic to filter by:</b>"),
        widgets.HBox([original_demo, original_value]),
        widgets.HTML("<b>Select predicted demographic to examine:</b>"),
        widgets.HBox([predicted_demo, predicted_value]),
        widgets.HBox([n_samples, show_button]),
        output
    ])
    
    return controls

# Create and display the explorer
explorer = create_demographic_explorer(data_rafdb)
display(explorer)
```
